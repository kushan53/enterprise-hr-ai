Department-level satisfaction aggregation, burnout risk detection, and flight-risk correlation.


### 1. Engagement Aggregations by Department


```python
import pandas as pd

df = pd.read_csv('../data/processed/employee_intelligence.csv')
eng_summary = df.groupby('Department').agg(
    TotalEmployees=('EmployeeID', 'count'),
    AvgEngagement=('EngagementScore', 'mean'),
    AvgWLB=('WorkLifeBalanceScore', 'mean'),
    HighRiskCount=('AttritionRiskCategory', lambda x: (x == 'High').sum())
).reset_index()

display(eng_summary)
```
