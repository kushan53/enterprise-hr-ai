Validation checks ensuring column existence, strict type adherence, range checks, and uniqueness assertions.


### 1. Schema Validation Setup


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/employee_attrition.csv')
print('Loaded rows:', len(df))
```

### 2. Type & Range Validation Assertions


```python
# Assert ID uniqueness
assert df['EmployeeID'].is_unique, 'Duplicate EmployeeID found!'

# Assert valid age range
assert df['Age'].between(18, 75).all(), 'Age out of reasonable bounds'

# Assert valid salary
assert (df['MonthlySalary'] > 0).all(), 'Monthly salary must be positive'

# Assert valid Attrition values
assert set(df['AttritionRisk'].unique()) <= {'Yes', 'No'}, 'Unexpected Attrition values'

print('All primary schema assertions passed successfully!')
```

### 3. Outlier and Null Boundary Scan


```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
print('Numerical Summary:')
display(df[numeric_cols].describe().T)
```
