Explaining predictive features across the organization and per-employee risk drivers using SHAP values.


### 1. Feature Importance & SHAP Drivers


```python
import json, pandas as pd
import matplotlib.pyplot as plt

with open('../models/v2/metadata.json') as f:
    v2_meta = json.load(f)

imp_df = pd.DataFrame(v2_meta['top_contributing_features'])
print('Top 10 Global Risk Drivers:')
display(imp_df)

plt.figure(figsize=(10, 5))
plt.barh(imp_df['feature'][::-1], imp_df['importance'][::-1], color='#3b82f6')
plt.title('Top Features Driving Flight Risk')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```
