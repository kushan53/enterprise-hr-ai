# 05. Feature Engineering for Predictive HR Analytics
**Enterprise HR AI - Workforce Intelligence & Upskilling Platform**

Creating high-signal domain features including tenure velocity, overtime ratio, promotion gap, and burnout risk index.


### 1. Engineering HR Signals


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/employees.csv')

# Feature Engineering
df['YearsSincePromotion'] = 2024 - pd.to_numeric(df['LastPromotionYear'], errors='coerce').fillna(2020)
df['SalaryPerYearAtCompany'] = df['MonthlySalary'] / (df['YearsAtCompany'] + 1)
df['OvertimeRatio'] = df['OvertimeHoursPerMonth'] / 160.0
df['BurnoutRiskIndex'] = (df['OvertimeHoursPerMonth'] * 0.4) - (df['WorkLifeBalanceScore'] * 10) + (df['ProjectsHandled'] * 2)

print('Engineered Features Head:')
display(df[['Name', 'OvertimeRatio', 'YearsSincePromotion', 'SalaryPerYearAtCompany', 'BurnoutRiskIndex']].head())
```
