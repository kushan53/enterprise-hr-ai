### 1. Setup and Environment


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
DATA_PATH = '../data/raw'
print('Raw Data Files:', os.listdir(DATA_PATH))
```

### 2. Loading and Inspecting Employee Attrition


```python
df_attrition = pd.read_csv(f'{DATA_PATH}/employee_attrition.csv')
print('Shape:', df_attrition.shape)
df_attrition.head()
```

```python
print('Info:')
print(df_attrition.info())
print('\nMissing Values:\n', df_attrition.isnull().sum())
print('\nTarget Distribution:\n', df_attrition['AttritionRisk'].value_counts(normalize=True) * 100)
```

### 3. Inspecting HR Performance & Engagement


```python
df_perf = pd.read_csv(f'{DATA_PATH}/Employee_Performance_Dataset.csv')
print('Performance shape:', df_perf.shape)
df_perf.head()
```

```python
df_eng = pd.read_csv(f'{DATA_PATH}/Cleaned_HR_Data_Analysis.csv')
print('Engagement shape:', df_eng.shape)
df_eng.head()
```
