# 09. Model Registry & Versioning Pipeline
**Enterprise HR AI - Workforce Intelligence & Upskilling Platform**

Managing versioned artifacts, metadata tracking, and deployment readiness under models/v1/ and models/v2/.


### 1. Registry Inspection


```python
import os, json
for ver in ['v1', 'v2']:
    meta_path = f'../models/{ver}/metadata.json'
    if os.path.exists(meta_path):
        with open(meta_path) as f:
            print(f'=== Registry: {ver} ===')
            print(json.dumps(json.load(f), indent=2))
```
