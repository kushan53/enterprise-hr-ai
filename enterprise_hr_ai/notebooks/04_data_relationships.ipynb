**Enterprise HR AI - Workforce Intelligence & Upskilling Platform**

Connecting employee records with role mappings, performance history, and skills taxonomy.


### 1. Loading Processed Tables


```python
import pandas as pd

df_emp = pd.read_csv('../data/processed/employees.csv')
df_roles = pd.read_csv('../data/processed/role_skills.csv')
df_skills = pd.read_csv('../data/processed/employee_skills.csv')

print(f'Employees: {len(df_emp)}, Role Skills: {len(df_roles)}, Employee Skills: {len(df_skills)}')
```

### 2. Validating Entity Relational Joins


```python
# Join employees with their skills
merged_skills = pd.merge(df_emp[['EmployeeID', 'Name', 'JobRole']], df_skills, on='EmployeeID')
print('Employee Skills Join Head:')
display(merged_skills.head(10))
```
