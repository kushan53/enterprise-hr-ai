# 03. Data Cleaning & Normalization
**Enterprise HR AI - Workforce Intelligence & Upskilling Platform**

Standardizing column names, imputing missing records, encoding targets, and exporting cleaned datasets to data/processed/.


### 1. Data Cleaning Pipeline


```python
import pandas as pd
import numpy as np
import os

df = pd.read_csv('../data/raw/employee_attrition.csv')
# Standardize column names
df.columns = [c.strip() for c in df.columns]

# Fill missing values
if 'CustomerSatisfaction' in df.columns:
    df['CustomerSatisfaction'] = df['CustomerSatisfaction'].fillna(df['CustomerSatisfaction'].median())

# Standardize target to binary int
df['Attrition'] = df['AttritionRisk'].map({'Yes': 1, 'No': 0})

os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/employee_attrition_processed.csv', index=False)
print('Cleaned dataset saved successfully.')
```
